```python
%%sql
USE COMPANY;
SELECT * FROM EMPLOYEES;
```

```python
%%sql
SELECT DEPARTMENT, ROUND(AVG(SALARY),0) AS AVG_SALARY
FROM EMPLOYEES 
GROUP BY DEPARTMENT;
```

```python
%%sql
SELECT DEPARTMENT, COUNT(E_ID) FROM 
EMPLOYEES
GROUP BY DEPARTMENT;
```

```python
%%sql
SELECT DEPARTMENT, COUNT(E_ID) FROM 
EMPLOYEES
GROUP BY DEPARTMENT
HAVING COUNT(E_ID)>1
ORDER BY DEPARTMENT;
```

```python
%%sql
SELECT DEPARTMENT, COUNT(E_ID)
FROM EMPLOYEES
WHERE SALARY >30000
GROUP BY DEPARTMENT
HAVING COUNT(E_ID)>1
ORDER BY DEPARTMENT;
```

```python
%%sql
SELECT DEPARTMENT , round(AVG(SALARY))
FROM EMPLOYEES
GROUP BY DEPARTMENT
HAVING AVG(SALARY) >=35000;
```
