```python
%%sql
CREATE DATABASE IF NOT EXISTS VIGNESH_SEKAR;
USE VIGNESH_SEKAR;
```

```python
%%sql
CREATE TABLE BRANCH ( 
B_ID INT PRIMARY KEY,
B_NAME VARCHAR(20),
STATE VARCHAR(20)
);
```

```python
%%sql
SELECT * FROM BRANCH;
```

```python
%%sql
CREATE TABLE EMPLOYEES (
E_ID INT PRIMARY KEY,
E_NAME VARCHAR(30),
DEPARTMENT VARCHAR (30),
SALARY DECIMAL(7,2),
B_ID INT,
CONSTRAINT FK_B_ID FOREIGN KEY (B_ID) REFERENCES BRANCH(B_ID)
);
```

```python
%%sql
DROP TABLE BRANCH;
```

```python
%%sql
-- CANNOT DROP BRANCH TABLE BECAUSE EMPLOYEES TABLE REFERENCES THE BRANCH TABLE
-- WE CAN DROP EMPLOYEES TABLE BEFORE DROPPING BRANCH TABLE TO AVOID ERRORS
```

```python
%%sql
DROP TABLE EMPLOYEES;
```

```python
%%sql
-- DROPPING FOREIGN KEY CONSTRAINT

ALTER TABLE EMPLOYEES
DROP FOREIGN KEY FK_B_ID; 
-- ONLY DROP LINK BETWEEN TWO TABLES NOT THE COLUMN

-- NOW WE CAN DROP BRANCH TABLE 
```
