```python
%%sql
SELECT * FROM EMPLOYEES
WHERE DEPARTMENT LIKE "%KET%"; 
```

```python
%%sql
SELECT * FROM EMPLOYEES
WHERE E_NAME LIKE "S%N";
```

```python
%%sql
SELECT * FROM EMPLOYEES
WHERE E_NAME LIKE "P%D%P";
```

```python
%%sql
--  SINGLE CHARACTER
SELECT * FROM EMPLOYEES
WHERE E_NAME LIKE "_I%";
```

```python
%%sql
SELECT * FROM EMPLOYEES
WHERE E_NAME LIKE "_I%" AND DEPARTMENT LIKE "%G";
```

```python
%%sql
SELECT * FROM EMPLOYEES
WHERE E_NAME LIKE "_I%" OR DEPARTMENT LIKE "%G";
```

```python
%%sql
SELECT * FROM EMPLOYEES
WHERE DEPARTMENT LIKE "F__A%";
```

```python
%%sql
-- BETWEEN 
SELECT * FROM EMPLOYEES
WHERE SALARY BETWEEN 30000 AND 75000;
```

```python
%%sql
SELECT E_NAME AS NAME, 
SALARY AS "MONTHLY SALARY",
DEPARTMENT AS DPT 
FROM EMPLOYEES;
```

```python
%%sql
SELECT SALARY, SALARY * 12 AS YEARLY_SALARY FROM EMPLOYEES 
WHERE SALARY >= 40000;
```
