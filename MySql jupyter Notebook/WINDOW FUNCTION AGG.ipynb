```python
%%sql
USE COMPANY;
```

```python
%%sql
CREATE TABLE SALES(
ID INT ,
BRANCH VARCHAR (30),
AMOUNT INT);
```

```python
%%sql
INSERT INTO SALES VALUES(1,"CHENNAI",400),
(2,"TIRUVALLUR",350),
(3,"VELLORE",200),
(4,"VELLORE",650),
(5,"CHENNAI",350),
(6,"CHENNAI",200),
(7,"TIRUVALLUR",250);
```

```python
%%sql
SELECT * FROM SALES;
```

```python
%%sql
SELECT BRANCH, 
SUM(AMOUNT) FROM SALES GROUP BY BRANCH
#HAVING SUM(AMOUNT)>600;
```

```python
%%sql
SELECT ID,BRANCH, 
SUM(AMOUNT) OVER () AS "TOTAL SALES" FROM SALES;
```

```python
%%sql
SELECT ID,BRANCH, 
SUM(AMOUNT) OVER (PARTITION BY BRANCH) AS "TOTAL SALES" FROM SALES;
```

```python
%%sql
-- AVG 
SELECT BRANCH, ROUND(AVG(AMOUNT)) FROM SALES GROUP BY BRANCH;
```
