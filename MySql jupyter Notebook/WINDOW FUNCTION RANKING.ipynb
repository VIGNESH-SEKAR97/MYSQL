```python
%%sql
USE COMPANY;
SELECT * FROM SALES;
```

```python
%%sql
-- ROW_NUMBER()
SELECT *,
ROW_NUMBER() OVER (ORDER BY AMOUNT DESC)  AS "ROW NUMBER" FROM SALES;
```

```python
%%sql
SELECT *,
ROW_NUMBER() OVER (PARTITION BY BRANCH)  AS "ROW NUMBER" FROM SALES;
```

```python
%%sql
SELECT *,
ROW_NUMBER() OVER (ORDER BY ID)  AS "ROW NUMBER" FROM SALES;
```

```python
%%sql
SELECT *,
ROW_NUMBER() OVER (PARTITION BY BRANCH ORDER BY AMOUNT DESC)  AS "ROW NUMBER" FROM SALES;
```

```python
%%sql
-- RANK()
SELECT *,
RANK() OVER (ORDER BY AMOUNT DESC)  AS "RANK" FROM SALES;
```

```python
%%sql
SELECT *,
RANK() OVER (ORDER BY BRANCH )  AS "RANK" FROM SALES;
```

```python
%%sql
SELECT *,
RANK() OVER (ORDER BY ID )  AS "RANK" FROM SALES;
```

```python
%%sql
-- DENSE_RANK()
SELECT *,
DENSE_RANK() OVER (ORDER BY AMOUNT)  AS "DENSE RANK" FROM SALES;
```

```python
%%sql
SELECT *,
DENSE_RANK() OVER (ORDER BY BRANCH)  AS "DENSE RANK" FROM SALES;
```

```python
%%sql
SELECT *,
DENSE_RANK() OVER (PARTITION BY BRANCH ORDER BY AMOUNT DESC)  AS "DENSE RANK" FROM SALES;
```

```python
%%sql
-- PERCENT_RANK()
-- RANK OF THE ROW -1 / TOTAL ROW - 1
```

```python
%%sql
SELECT *,
PERCENT_RANK() OVER (ORDER BY AMOUNT DESC )  AS "PERCENT RANK" FROM SALES;
```

```python
%%sql
-- NTILE_RANK()

SELECT *,
NTILE(3) OVER (ORDER BY AMOUNT DESC )  AS "PERCENT RANK" FROM SALES;
```
