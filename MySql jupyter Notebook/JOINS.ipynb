```python
%%sql
USE COMPANY;
SELECT * FROM EMPLOYEES;
```

```python
%%sql
SELECT * FROM BRANCH;
```

```python
%%sql
-- INNER JOINS
SELECT E.E_NAME, E.DEPARTMENT,B.B_NAME
FROM EMPLOYEES AS E
INNER JOIN BRANCH AS B
ON E.B_ID = B.B_ID;
```

```python
%%sql
-- LEFT JOINS
SELECT E.E_NAME, E.DEPARTMENT,B.B_NAME
FROM EMPLOYEES AS E
LEFT JOIN BRANCH AS B
ON E.B_ID = B.B_ID;
```

```python
%%sql
-- RIGHT JOINS
SELECT E.E_NAME, E.DEPARTMENT,B.B_NAME
FROM EMPLOYEES AS E
RIGHT JOIN BRANCH AS B
ON E.B_ID = B.B_ID;
```

```python
%%sql
-- CROSS JOINS
SELECT E.E_NAME, E.DEPARTMENT,B.B_NAME
FROM EMPLOYEES AS E
CROSS JOIN BRANCH AS B
ON E.B_ID = B.B_ID;
```

```python
%%sql
-- UNON AND UNION ALL
-- NUMBER OF COLUMNS SHOULD MATCH TO GET OUTPUT 
-- UNION GAVE MATCHED RESULT ONLY WHEREAS UNION ALL GIVE ALL THE RECORDS
```

```python
%%sql
SELECT * FROM EMPLOYEES
UNION ALL
SELECT * FROM BRANCH;
```
