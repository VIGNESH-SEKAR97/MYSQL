```python
%%sql
CREATE TABLE EMPLOYEES (
E_ID INT PRIMARY KEY,
E_NAME VARCHAR(30),
DEPARTMENT VARCHAR (30),
SALARY DECIMAL(7,2),
EMAIL VARCHAR(30),
PH_NUMBER VARCHAR(10)
);
```

```python
%%sql
-- TO READ TO THE TABLE USE SELECT QUERY 
SELECT * FROM EMPLOYEES;
```

```python
%%sql
-- DROP TABLE
DROP TABLE EMPLOYEES;
```

```python
%%sql
-- TO ADD A NEW COLUMN
ALTER TABLE EMPLOYEES ADD BRANCH_ID VARCHAR (2);
```

```python
%%sql
SELECT * FROM EMPLOYEES;
```

```python
%%sql
-- TO DROP A COLUMN
ALTER TABLE EMPLOYEES DROP BRANCH_ID;
```

```python
%%sql
SELECT * FROM EMPLOYEES;
```

```python
%%sql
ALTER TABLE EMPLOYEES
MODIFY SALARY DECIMAL(7,0);
```

```python
%%sql
ALTER TABLE EMPLOYEES
DROP EMAIL;
```
