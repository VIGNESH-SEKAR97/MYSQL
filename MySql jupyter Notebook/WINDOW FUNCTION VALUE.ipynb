```python
%%sql
USE COMPANY;
CREATE TABLE WORKERS(
ID INT ,
NAME VARCHAR (30),
SALARY INT);
```

```python
%%sql
SELECT * FROM WORKERS;
```

```python
%%sql
-- LAG
SELECT *,
LAG(SALARY) OVER (PARTITION BY ID ORDER BY SALARY) AS INCREMENTED_SALARY
FROM WORKERS;
```

```python
%%sql
-- LAG
SELECT *,
LAG(SALARY) OVER (PARTITION BY NAME ORDER BY SALARY) AS INCREMENTED_SALARY
FROM WORKERS;
```

```python
%%sql
-- LEAD
SELECT *,
LEAD(SALARY) OVER (PARTITION BY NAME ORDER BY SALARY) AS INCREMENTED_SALARY
FROM WORKERS;
```

```python
%%sql
-- FIRST_VALUE
SELECT *,
FIRST_VALUE(SALARY) OVER (PARTITION BY ID ORDER BY SALARY) AS FIRST_SALARY
FROM WORKERS;
```

```python
%%sql
SELECT *,
    LAST_VALUE(SALARY) OVER (PARTITION BY ID ORDER BY SALARY
     ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING ) AS LAST_SALARY
FROM WORKERS;
```

```python
%%sql
SELECT *,
    LAST_VALUE(SALARY) OVER (PARTITION BY NAME ORDER BY SALARY
     ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING ) AS LAST_SALARY
FROM WORKERS;
```

```python
%%sql
SELECT *,
NTH_VALUE(SALARY, 2) OVER (PARTITION BY ID ORDER BY SALARY
        ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING) AS SECOND_SALARY
FROM WORKERS;
```

```python
%%sql
SELECT *,
NTH_VALUE(SALARY, 2) OVER (PARTITION BY NAME ORDER BY SALARY
        ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING) AS SECOND_SALARY
FROM WORKERS;
```

```python
%%sql
SELECT *,
    SALARY - LAG(SALARY) OVER (PARTITION BY ID ORDER BY SALARY) AS SALARY_DIFFERENCE
FROM WORKERS;
```
